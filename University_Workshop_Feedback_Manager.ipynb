```python
feedback = " THE SPEAKER WAS GREAT but THE ROOM WAS COLD "
cleaned = feedback.strip().lower()
cleaned = cleaned.replace("speaker", "presenter")
cleaned = " ".join(cleaned.split())
cleaned = cleaned.title()

print(f"Cleaned feedback: {cleaned}")
```

```python
feedback_list = [
    "The Presenter Was Great But The Room Was Cold",
    "The Workshop Was Informative And Well Organized",
    "I Enjoyed The Practical Examples"
]

with open("feedback.txt", "w") as file:
    for feedback in feedback_list:
        file.write(feedback + "\n")

print("Initial feedback:")
with open("feedback.txt", "r") as file:
    for line in file:
        print(line.strip())

with open("feedback.txt", "a") as file:
    file.write("The Presenter Explained The Topic Clearly\n")

print("\nUpdated feedback:")
with open("feedback.txt", "r") as file:
    for line in file:
        print(line.strip())
```

```python
try:
    with open("feedback.txt", "r") as file:
        for line in file:
            print(line.strip())
except FileNotFoundError:
    print("File not found. Please create feedback.txt first.")
except PermissionError:
    print("Permission denied. Close the file and try again.")
finally:
    print("Operation completed.")
```

```python
with open("feedback.txt", "r") as file:
    lines = file.readlines()

total_feedback = len(lines)
great_count = 0

for line in lines:
    if "great" in line.lower():
        great_count += 1

summary = f"""=== Workshop Feedback Summary ===
Total Feedback: {total_feedback}
Mentions of 'Great': {great_count}"""

with open("summary.txt", "w") as file:
    file.write(summary)

print(summary)
```

```python
feedback = " THE SPEAKER WAS GREAT but THE ROOM WAS COLD "
cleaned = feedback.strip().lower()
cleaned = cleaned.replace("speaker", "presenter")
cleaned = " ".join(cleaned.split())
cleaned = cleaned.title()

print(f"Cleaned feedback: {cleaned}")

feedback_list = [
    cleaned,
    "The Workshop Was Informative And Well Organized",
    "I Enjoyed The Practical Examples"
]

with open("feedback.txt", "w") as file:
    for message in feedback_list:
        file.write(message + "\n")

print("\nInitial feedback:")
with open("feedback.txt", "r") as file:
    for line in file:
        print(line.strip())

with open("feedback.txt", "a") as file:
    file.write("The Presenter Explained The Topic Clearly\n")

print("\nUpdated feedback:")
try:
    with open("feedback.txt", "r") as file:
        for line in file:
            print(line.strip())
except FileNotFoundError:
    print("File not found. Please create feedback.txt first.")
except PermissionError:
    print("Permission denied. Close the file and try again.")
finally:
    print("Operation completed.")

with open("feedback.txt", "r") as file:
    lines = file.readlines()

total_feedback = len(lines)
great_count = 0

for line in lines:
    if "great" in line.lower():
        great_count += 1

summary = f"""=== Workshop Feedback Summary ===
Total Feedback: {total_feedback}
Mentions of 'Great': {great_count}"""

with open("summary.txt", "w") as file:
    file.write(summary)

print(f"\n{summary}")
```
