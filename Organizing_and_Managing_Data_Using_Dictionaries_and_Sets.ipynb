```python
shipments = {
    "SHP101": "Delivered",
    "SHP102": "Pending",
    "SHP103": "In Transit",
    "SHP104": "Delivered"
}

shipment_id = "SHP103"

print("Shipment:", shipment_id)
print("Status:", shipments[shipment_id])
```

```python
shipments = {
    "SHP101": "Delivered",
    "SHP102": "Pending",
    "SHP103": "In Transit"
}

shipments["SHP102"] = "Delivered"

shipments["SHP104"] = "Pending"

del shipments["SHP101"]

print(shipments)
```

```python
print(shipments.keys())
```

```python
print(shipments.values())
```

```python
for shipment_id, status in shipments.items():
    print(shipment_id, "-", status)
```

```python
north = {"New York", "Dallas", "Miami"}

south = {"Miami", "Atlanta", "Dallas"}

all_cities = north.union(south)

print(all_cities)
```

```python
common_cities = north.intersection(south)

print(common_cities)
```

```python
north_only = north.difference(south)

print(north_only)
```
